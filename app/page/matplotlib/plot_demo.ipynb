```python
import random

import matplotlib.pyplot as plt

# SimHei 是 Windows 自带字体，macOS 上没有；按顺序取第一个可用的中文字体
plt.rcParams['font.sans-serif'] = [
    'Heiti SC', 'Microsoft YaHei', 'SimHei', 'Arial Unicode MS', 'sans-serif',
]


def drawn_line():
    x = [e for e in range(1, 22)]
    y_bj = [random.randint(10, 16) for i in x]
    y_xa = [random.randint(13, 18) for i in x]

    plt.figure(figsize=(10, 5))
    plt.plot(x, y_bj, label="背景")
    plt.plot(x, y_xa, label="西安")
    plt.xlabel('时间')
    plt.ylabel('气温')

    y_ticks = [e for e in range(min([*y_bj, *y_xa]) - 3, max([*y_bj, *y_xa]) + 3)]
    plt.yticks(y_ticks)

    plt.title("气温变化图", fontsize=20)

    plt.grid(True, linestyle='--', alpha=0.3)
    plt.legend()
    plt.show()
```

```python
from matplotlib.axes import Axes


def drawn_muti_char():
    figure, axes = plt.subplots(nrows=1, ncols=2, figsize=(20, 6), dpi=100)

    countries = ["中国", "美国", "印度", "加拿大", "伊拉克", "沙特", "伊朗"]
    values = [35, 23, 18, 21, 56, 78, 51]
    axes0: Axes = axes[0]
    axes0.bar(countries, values, width=0.6, color="green")
    axes0.set_title("世界石油储备")
    axes0.set_xlabel("国家")
    axes0.set_ylabel("石油储备(亿吨)")
    axes0.grid(linestyle='--', alpha=0.3)

    countries2 = ['印度', '中国', '美国', '印尼', '巴基斯坦', '尼日利亚', '巴西', '俄罗斯', '其他']
    values2 = [14.51, 14.09, 3.4, 2.83, 2.51, 2.33, 2.12, 1.44, 20]

    axes1: Axes = axes[1]
    axes1.set_title("世界石油储备占比")
    axes1.pie(values2, labels=countries2, autopct="%1.1f%%", )
    axes1.legend(loc="lower center", ncol=5, bbox_to_anchor=(0.5, -0.1))

    plt.show()


drawn_muti_char()
```
